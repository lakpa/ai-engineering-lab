## 1. Learning Goal

**Does fine tuning classify our messages more reliably than prompting alone?**

The previous notebook used RAG to give a language model song descriptions and lyrics so it could explain a recommendation. Now imagine that the music service receives messages such as “Why was this song recommended?” and “My playlist disappeared.” Before answering, the application needs to understand what each sender wants and choose the right workflow.

We will explore classifying these messages into categories specific to the service, including recommendation requests, explanation requests, customer support, partnership proposals, junk, and human review. We will first try instructions and examples in a prompt, then compare that approach with a model given additional training on labelled messages.

RAG supplies information for an answer. Fine tuning updates a model's parameters to encourage a desired behavior, such as choosing a category. This experiment asks whether that extra training improves classification on unseen messages enough to justify the work.
